```python
import numpy as np

# Sigmoid activation function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Input dataset (XOR)
X = np.array([[0, 0],
              [0, 1],
              [1, 0],
              [1, 1]])

Y = np.array([[0], [1], [1], [0]])

# Initialize weights
np.random.seed(42)
W1 = np.random.rand(2, 4)
W2 = np.random.rand(4, 1)

b1 = np.zeros((1, 4))
b2 = np.zeros((1, 1))

lr = 0.5

# Training using Backpropagation
for epoch in range(10000):

    # Forward propagation
    H = sigmoid(np.dot(X, W1) + b1)
    O = sigmoid(np.dot(H, W2) + b2)

    # Calculate error
    error = Y - O

    # Backpropagation
    dO = error * O * (1 - O)
    dH = np.dot(dO, W2.T) * H * (1 - H)

    # Update weights
    W2 += lr * np.dot(H.T, dO)
    W1 += lr * np.dot(X.T, dH)

    b2 += lr * np.sum(dO, axis=0)
    b1 += lr * np.sum(dH, axis=0)

# Display results
print("Input:")
print(X)

print("\nActual Output:")
print(Y)

print("\nPredicted Output:")
print(np.round(O, 2))
```
