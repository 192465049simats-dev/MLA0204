```python
import pandas as pd

data = pd.DataFrame([
    ['sunny', 'warm', 'normal', 'strong', 'warm', 'same', 'yes'],
    ['sunny', 'warm', 'high',   'strong', 'warm', 'same', 'yes'],
    ['rainy', 'cold', 'high',   'strong', 'warm', 'change', 'no'],
    ['sunny', 'warm', 'high',   'strong', 'cool', 'change', 'yes']
], columns=['sky', 'airtemp', 'humidity', 'wind',
            'water', 'forecast', 'enjoysport'])

def find_s(data):
    hypothesis = ['0'] * 6

    for _, row in data.iterrows():
        if row['enjoysport'] == 'yes':
            for i in range(6):
                if hypothesis[i] == '0':
                    hypothesis[i] = row.iloc[i]
                elif hypothesis[i] != row.iloc[i]:
                    hypothesis[i] = '?'

    return hypothesis

h = find_s(data)

print("Most Specific Hypothesis:")
print(h)
```

```python
import pandas as pd

data = pd.DataFrame([
    ['sunny', 'warm', 'normal', 'strong', 'warm', 'same', 'yes'],
    ['sunny', 'warm', 'high',   'strong', 'warm', 'same', 'yes'],
    ['rainy', 'cold', 'high',   'strong', 'warm', 'change', 'no'],
    ['sunny', 'warm', 'high',   'strong', 'cool', 'change', 'yes']
], columns=['sky', 'airtemp', 'humidity', 'wind',
            'water', 'forecast', 'enjoysport'])

attributes = data.columns[:-1]
n = len(attributes)

# Initial boundaries
S = ['0'] * n
G = [['?'] * n]


def covers(h, example):
    for i in range(n):
        if h[i] != '?' and h[i] != example[i]:
            return False
    return True


def more_general(h1, h2):
    for i in range(n):
        if h1[i] != '?' and h1[i] != h2[i]:
            return False
    return True


for _, row in data.iterrows():

    example = list(row.iloc[:n])
    target = row.iloc[n]

    if target == 'yes':

        # Remove G hypotheses that do not cover positive example
        G = [g for g in G if covers(g, example)]

        # Generalize S
        for i in range(n):
            if S[i] == '0':
                S[i] = example[i]
            elif S[i] != example[i]:
                S[i] = '?'

    else:

        # Remove G hypotheses covering negative example
        new_G = []

        for g in G:
            if covers(g, example):
                for i in range(n):
                    if g[i] == '?':
                        if S[i] != example[i]:
                            new_h = g.copy()
                            new_h[i] = S[i]

                            if new_h not in new_G:
                                new_G.append(new_h)
            else:
                new_G.append(g)

        G = new_G

print("Specific Boundary (S):")
print(S)

print("\nGeneral Boundary (G):")
for g in G:
    print(g)
```
