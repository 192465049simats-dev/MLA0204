```python
import numpy as np
from sklearn.mixture import GaussianMixture

# Dataset
X = np.array([[1], [2], [3], [8], [9], [10]])

# Create Gaussian Mixture Model
model = GaussianMixture(
    n_components=2,
    random_state=42
)

# Train using EM algorithm
model.fit(X)

# Predict clusters
labels = model.predict(X)

# Display results
print("Dataset:")
print(X.flatten())

print("\nCluster Labels:")
print(labels)

print("\nCluster Means:")
print(model.means_.flatten())

print("\nConverged:", model.converged_)
```
