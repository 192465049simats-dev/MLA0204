```python

import pandas as pd
from sklearn.ensemble import RandomForestRegressor

# Sample dataset
data = {
    'RAM': [2, 3, 4, 6, 8, 12, 4, 6],
    'Storage': [32, 32, 64, 128, 128, 256, 64, 128],
    'Battery': [3000, 3500, 4000, 4500, 5000, 6000, 4200, 4800],
    'Price': [8000, 10000, 12000, 18000, 25000, 40000, 14000, 22000]
}

df = pd.DataFrame(data)

X = df[['RAM', 'Storage', 'Battery']]
y = df['Price']

# Train model
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X, y)

# Predict mobile price
new_mobile = [[8, 128, 5000]]
prediction = model.predict(new_mobile)

print("Mobile Specifications:", new_mobile)
print("Predicted Mobile Price: Rs.", round(prediction[0], 2))
```
