```python

import pandas as pd
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import LabelEncoder

# Create dataset
data = {
    'Income': [25000, 40000, 50000, 60000, 30000,
               70000, 45000, 80000],
    'Credit_Score': [550, 650, 700, 750, 580,
                     800, 680, 820],
    'Age': [25, 30, 35, 40, 28, 45, 32, 50],
    'Loan_Status': ['No', 'No', 'Yes', 'Yes',
                    'No', 'Yes', 'Yes', 'Yes']
}

df = pd.DataFrame(data)

X = df[['Income', 'Credit_Score', 'Age']]
y = df['Loan_Status']

# Train Naive Bayes model
model = GaussianNB()
model.fit(X, y)

# Predict loan status
new_customer = [[55000, 720, 35]]
prediction = model.predict(new_customer)

print("Customer Details:", new_customer)
print("Loan Prediction:", prediction[0])
```
