```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder

# Create dataset
data = {
    'Income': [25000, 30000, 50000, 60000, 20000, 70000],
    'Age': [22, 25, 35, 40, 21, 45],
    'Loan': [15000, 20000, 10000, 15000, 18000, 12000],
    'Credit_Score': ['Low', 'Low', 'High', 'High', 'Low', 'High']
}

df = pd.DataFrame(data)

# Split features and target
X = df[['Income', 'Age', 'Loan']]
y = df['Credit_Score']

# Create model
model = DecisionTreeClassifier(random_state=42)

# Train model
model.fit(X, y)

# Test new customer
new_customer = pd.DataFrame({
    'Income': [55000],
    'Age': [38],
    'Loan': [12000]
})

prediction = model.predict(new_customer)

print("New Customer Details:")
print(new_customer)

print("\nPredicted Credit Score:", prediction[0])
```
