```python
import numpy as np
from sklearn.ensemble import RandomForestRegressor

# Dataset
X = np.array([
    [1000, 2],
    [1500, 3],
    [2000, 3],
    [2500, 4],
    [3000, 4],
    [3500, 5]
])

y = np.array([20, 30, 40, 50, 60, 70])

# Create model
model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

# Train model
model.fit(X, y)

# Predict house price
new_house = np.array([[2200, 3]])

prediction = model.predict(new_house)

print("House Details:")
print("Area: 2200 sq.ft")
print("Bedrooms: 3")

print("\nPredicted House Price:",
      round(prediction[0], 2), "Lakhs")
```
