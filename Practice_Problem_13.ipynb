```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Dataset
X = np.array([
    [2, 20000],
    [3, 30000],
    [4, 40000],
    [5, 50000],
    [6, 60000]
])

y = np.array([8, 7, 6, 5, 4])

# Create model
model = LinearRegression()

# Train model
model.fit(X, y)

# Predict car price
new_car = np.array([[3, 25000]])

prediction = model.predict(new_car)

print("Car Details:")
print("Age: 3 years")
print("Kilometers: 25000")

print("\nPredicted Car Price:",
      round(prediction[0], 2), "Lakhs")
```
